# Amazon Books & Movies Data Filtering

Loads the first 100 lines of each reviews/metadata file (gzipped JSONL) into a DataFrame and previews them.

```python
import gzip
import json
from pathlib import Path

import pandas as pd

DATA_DIR = Path("../data/raw")

N_LINES = 100


def load_jsonl_head(path: Path, n: int = N_LINES) -> pd.DataFrame:
    records = []
    with gzip.open(path, "rt", encoding="utf-8") as f:
        for i, line in enumerate(f):
            if i >= n:
                break
            records.append(json.loads(line))
    return pd.DataFrame(records)
```

# Find or load `user_id` overlap
This cell identifies the users that appear in both, then saves the overlap to a .pkl.

```python
import gzip
import json
import pickle
from tqdm import tqdm
from pathlib import Path

CACHE_DIR = Path("../data/cache")
CACHE_DIR.mkdir(exist_ok=True)

def stream_user_ids(path, desc=None) -> set:
    user_ids = set()
    with gzip.open(path, "rt", encoding="utf-8") as f:
        for line in tqdm(f, desc=desc, unit=" lines"):
            user_ids.add(json.loads(line)["user_id"])
    return user_ids

def load_or_compute(cache_path: Path, compute_fn):
    if cache_path.exists():
        with open(cache_path, "rb") as f:
            return pickle.load(f)
    result = compute_fn()
    with open(cache_path, "wb") as f:
        pickle.dump(result, f)
    return result

try:
    with open(CACHE_DIR / "common_user_ids.pkl", "rb") as f:
        common_user_ids_full = pickle.load(f)
except:
    book_user_ids_full = load_or_compute(
        CACHE_DIR / "book_user_ids.pkl",
        lambda: stream_user_ids(DATA_DIR / "books" / "Books.jsonl.gz", desc="Books reviews"),
    )
    movie_user_ids_full = load_or_compute(
        CACHE_DIR / "movie_user_ids.pkl",
        lambda: stream_user_ids(DATA_DIR / "movies" / "Movies_and_TV.jsonl.gz", desc="Movies reviews"),
    )

    print(f"Book reviewers: {len(book_user_ids_full):,}")
    print(f"Movie reviewers: {len(movie_user_ids_full):,}")
    common_user_ids_full = book_user_ids_full & movie_user_ids_full

    del book_user_ids_full               # for memory management
    del movie_user_ids_full              # for memory management

    with open(CACHE_DIR / "common_user_ids.pkl", "wb") as f:
        pickle.dump(common_user_ids_full, f)


print(f"Common reviewers: {len(common_user_ids_full):,}")
```

# Filter review datasets to common reviewers only

```python
import gzip
import json
from pathlib import Path

import pandas as pd
import pyarrow as pa
import pyarrow.parquet as pq
from tqdm import tqdm

PARQUET_DIR = Path("../data/filtered/")
PARQUET_DIR.mkdir(exist_ok=True)

REVIEW_SCHEMA = pa.schema([
    ("rating", pa.float32()),
    ("title", pa.string()),
    ("text", pa.string()),
    ("n_images", pa.int32()),
    ("asin", pa.string()),
    ("parent_asin", pa.string()),
    ("user_id", pa.string()),
    ("timestamp", pa.int64()),
    ("helpful_vote", pa.int32()),
    ("verified_purchase", pa.bool_()),
])
JSON_FIELDS = [f for f in REVIEW_SCHEMA.names if f != "n_images"]  # n_images is derived
BATCH_SIZE = 100_000

def filter_reviews_to_parquet(src, dst, keep_user_ids, desc=None, batch_size=BATCH_SIZE):
    """Stream src, keep rows whose user_id is in keep_user_ids, write Parquet to dst.

    Returns the Path. Nothing is loaded into memory — call pd.read_parquet on the
    result when you actually want data.
    """
    if dst.exists():
        n_rows = pq.ParquetFile(dst).metadata.num_rows
        print(f"{dst.name}: already built — {n_rows:,} rows, "
              f"{dst.stat().st_size / 1e9:.2f} GB (skipping rebuild)")
        return dst

    tmp = dst.with_name(dst.name + ".tmp")
    writer = pq.ParquetWriter(tmp, REVIEW_SCHEMA, compression="zstd", compression_level=3)
    buf = []
    kept = total = 0

    def flush():
        nonlocal buf
        if buf:
            writer.write_table(pa.Table.from_pylist(buf, schema=REVIEW_SCHEMA))
            buf = []

    try:
        with gzip.open(src, "rt", encoding="utf-8") as f:
            for line in tqdm(f, desc=desc, unit=" lines"):
                total += 1
                rec = json.loads(line)
                if rec.get("user_id") not in keep_user_ids:
                    continue
                row = {k: rec.get(k) for k in JSON_FIELDS}
                row["n_images"] = len(rec.get("images") or [])
                buf.append(row)
                kept += 1
                if len(buf) >= batch_size:
                    flush()
        flush()
    finally:
        writer.close()

    tmp.replace(dst)
    print(f"{dst.name}: kept {kept:,} of {total:,} rows ({kept / total:.2%})")
    print(f"  {src.name} {src.stat().st_size / 1e9:.2f} GB gzip"
          f"  ->  {dst.name} {dst.stat().st_size / 1e9:.2f} GB parquet")
    return dst


books_pq = filter_reviews_to_parquet(
    DATA_DIR / "books" / "Books.jsonl.gz",
    PARQUET_DIR / "books_reviews_common.parquet",
    common_user_ids_full,
    desc="Books reviews",
)

movies_pq = filter_reviews_to_parquet(
    DATA_DIR / "movies" / "Movies_and_TV.jsonl.gz",
    PARQUET_DIR / "movies_reviews_common.parquet",
    common_user_ids_full,
    desc="Movies reviews",
)
```

# Filter metadata

```python
import pyarrow.compute as pc

def parent_asins_from_parquet(path) -> set:
    # Reads only the parent_asin column — `text` never leaves disk.
    tbl = pq.read_table(path, columns=["parent_asin"])
    return set(pc.unique(tbl.column("parent_asin")).to_pylist())


books_asins = parent_asins_from_parquet(books_pq)
movies_asins = parent_asins_from_parquet(movies_pq)

print(f"Books:  {len(books_asins):,} distinct parent_asin")
print(f"Movies: {len(movies_asins):,} distinct parent_asin")
```

```python
META_SCHEMA = pa.schema([
    ("parent_asin", pa.string()),
    ("main_category", pa.string()),
    ("title", pa.string()),
    ("subtitle", pa.string()),
    ("author_name", pa.string()),
    ("average_rating", pa.float32()),
    ("rating_number", pa.int32()),
    ("price", pa.float32()),
    ("features", pa.list_(pa.string())),
    ("description", pa.list_(pa.string())),
    ("categories", pa.list_(pa.string())),
    ("store", pa.string()),
    ("details_json", pa.string()),
    ("n_images", pa.int32()),
    ("n_videos", pa.int32()),
])
META_BATCH_SIZE = 50_000


def _coerce_price(v):
    """price is float | None | str. Strings are '—' (no price) or 'from 4.95' (range floor)."""
    if isinstance(v, (int, float)):
        return float(v)
    if isinstance(v, str):
        try:
            return float(v.removeprefix("from ").strip())
        except ValueError:
            return None
    return None


def _meta_row(rec):
    author = rec.get("author")
    details = rec.get("details")
    return {
        "parent_asin": rec.get("parent_asin"),
        "main_category": rec.get("main_category"),
        "title": rec.get("title"),
        "subtitle": rec.get("subtitle"),
        "author_name": author.get("name") if isinstance(author, dict) else None,
        "average_rating": rec.get("average_rating"),
        "rating_number": rec.get("rating_number"),
        "price": _coerce_price(rec.get("price")),
        "features": rec.get("features") or [],
        "description": rec.get("description") or [],
        "categories": rec.get("categories") or [],
        "store": rec.get("store"),
        "details_json": json.dumps(details, ensure_ascii=False) if isinstance(details, dict) else None,
        "n_images": len(rec.get("images") or []),
        "n_videos": len(rec.get("videos") or []),
    }


def _read_cached_stats(dst):
    """kept + total from the Parquet footer. Footer-only read — touches no row data."""
    md = pq.ParquetFile(dst).metadata
    kv = md.metadata or {}
    total = int(kv[b"source_total_rows"]) if b"source_total_rows" in kv else None
    return md.num_rows, total


def _print_filter_stats(src, dst, kept, total, n_requested, cached, unit="parent_asins"):
    tag = "  (cached)" if cached else ""
    if total is None:
        print(f"{dst.name}: kept {kept:,} rows — source total not recorded{tag}")
    else:
        print(f"{dst.name}: kept {kept:,} of {total:,} rows ({kept / total:.2%}){tag}")
    print(f"  coverage: {kept:,} of {n_requested:,} requested {unit} "
          f"found in metadata ({kept / n_requested:.2%})")
    print(f"  {src.name} {src.stat().st_size / 1e9:.2f} GB gzip"
          f"  ->  {dst.name} {dst.stat().st_size / 1e9:.2f} GB parquet")


def filter_meta_to_parquet(src, dst, keep_parent_asins, desc=None, batch_size=META_BATCH_SIZE):
    """Stream src, keep rows whose parent_asin is in keep_parent_asins, write Parquet to dst.

    Returns the Path — nothing is loaded into memory.
    """
    if dst.exists():
        kept, total = _read_cached_stats(dst)
        cached = True
    else:
        cached = False
        tmp = dst.with_name(dst.name + ".tmp")
        writer = pq.ParquetWriter(tmp, META_SCHEMA, compression="zstd", compression_level=3)
        buf = []
        kept = total = 0

        def flush():
            nonlocal buf
            if buf:
                writer.write_table(pa.Table.from_pylist(buf, schema=META_SCHEMA))
                buf = []

        try:
            with gzip.open(src, "rt", encoding="utf-8") as f:
                for line in tqdm(f, desc=desc, unit=" lines"):
                    total += 1
                    rec = json.loads(line)
                    if rec.get("parent_asin") not in keep_parent_asins:
                        continue
                    buf.append(_meta_row(rec))
                    kept += 1
                    if len(buf) >= batch_size:
                        flush()
            flush()
            # Stamp the source line count into the footer so the cached path can report
            # it without gunzipping the source again.
            writer.add_key_value_metadata({
                "source_file": src.name,
                "source_total_rows": str(total),
            })
        finally:
            writer.close()
        tmp.replace(dst)

    _print_filter_stats(src, dst, kept, total, len(keep_parent_asins), cached)
    return dst


books_meta_pq = filter_meta_to_parquet(
    DATA_DIR / "books" / "meta_Books.jsonl.gz",
    PARQUET_DIR / "books_meta_common.parquet",
    books_asins,
    desc="Books metadata",
)

movies_meta_pq = filter_meta_to_parquet(
    DATA_DIR / "movies" / "meta_Movies_and_TV.jsonl.gz",
    PARQUET_DIR / "movies_meta_common.parquet",
    movies_asins,
    desc="Movies metadata",
)
```
